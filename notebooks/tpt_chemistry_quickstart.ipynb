# tpt-chemistry from Python

Hartree–Fock and SMILES helpers through the `tpt_chem` extension module.

Install the wheel first: `maturin build --release` in `crates/tpt-chem-py`, then `pip install` the file it writes to `target/wheels/`.

```python
import tpt_chem

print(tpt_chem.rhf_energy_xyz.__doc__)
```

## Energy from an XYZ file

Coordinates are in Å and the energy is in Hartree. This file is the water geometry from `data/`.

```python
from pathlib import Path

xyz = Path('../data/water.xyz').read_text()
energy = tpt_chem.rhf_energy_xyz(xyz)
print(f'E(RHF/STO-3G) = {energy:.8f} Eh')
```

## SMILES

Equivalent spellings of one molecule share a canonical form.

```python
for smiles in ['OC(C)=O', 'O=C(O)C', 'CC(=O)O']:
    print(f'{smiles:10s} -> {tpt_chem.canonical_smiles(smiles)}  ({tpt_chem.formula_from_smiles(smiles)})')
```
